```python
import numpy as np
import pandas as pds
import matplotlib.pyplot as plt
from matplotlib import cm


file =('/Users/shubhamjamdade/Desktop/Research/Oxygen_Helium Separation/Screening/Final Data/All Data.xlsx')
Data = pds.read_excel(file, 'Post Processing' )
Selectivity = np.array(Data['Avg_selectivity'])
Capacity = np.array(Data['Capacity'])
LCD = np.array(Data['LCD'])
PLD = np.array(Data['PLD'])
VF = np.array(Data['VF'])
Delta_H = np.array(Data['Delta_H'])
API = np.array(Data['API'])
T_opt = np.array(Data ['Temperature'])
OMS = np.array(Data ['OMS'])
GSA = np.array(Data ['GSA'])
```
