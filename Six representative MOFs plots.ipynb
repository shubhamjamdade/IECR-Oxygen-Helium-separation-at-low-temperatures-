```python
import numpy
import pandas as pds
import matplotlib.pyplot as plt
from matplotlib import cm
import seaborn as sns


file =('/Users/shubhamjamdade/Desktop/Paper O2_He/New - Data Processing/SIX MOFs Data Processing/Selectivity_Capacity_Regenerability.xlsx')
Data = pds.read_excel(file, 'Sequential Raw data' )

Selectivity =numpy.array(Data['Avg_selectivity'])
Capacity = numpy.array(Data['Capacity'])
Regenerability = numpy.array(Data['Regenerability'])
T = numpy.array(Data ['Temperature'])
T_K = numpy.array(Data ['T_K'])
MOF = numpy.array(Data ['MOF'])
X = numpy.array(Data['MolFraction'])
Delta_H = numpy.array(Data['Avg_Delta_H'])
```

```python
plt.scatter(x = Capacity, y = Regenerability, s=100*X, c=T, cmap="jet", edgecolor = 'black')
cbar = plt.colorbar(orientation="vertical", pad=0.02, shrink=1, aspect=25, format="%.3f")
cbar.set_label(label=r'$\ T (K)$', size=18, labelpad= 10)
cbar.set_ticks([100, 120, 140, 160, 180, 200])
cbar.set_ticklabels(["100","120","140", "160", "180", "200" ])
cbar.ax.tick_params(labelsize=15)

plt.clim(100,200)


plt.ylabel(r'$\ Re_{O_{2}}$', fontsize=18, labelpad= 10)
plt.xlabel(r'$\Delta N_{O_{2}} $', fontsize=18, labelpad= 10)

plt.xticks(numpy.arange(0, max(Capacity), 3.0))

plt.yticks(numpy.arange(0, max(Regenerability), 10.0))
plt.grid(False) 
plt.figure(dpi=10000)
plt.show()
```
